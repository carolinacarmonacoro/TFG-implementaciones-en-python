### Visualización tipos de aprendizaje

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

def draw_ml_tree_final():
    fig, ax = plt.subplots(figsize=(12, 9))
    ax.set_xlim(0, 110)
    ax.set_ylim(25, 100)
    ax.set_aspect('equal')
    ax.axis('off')

    # Estilos
    root_style = {'facecolor': "#2b77ad", 'edgecolor': 'black', 'linewidth': 1.5}
    cat_style = {'facecolor': "#f4df9a", 'edgecolor': 'black', 'linewidth': 1.2}
    sub_style = {'facecolor': "#f0c9ee", 'edgecolor': 'black', 'linewidth': 0.8}
    line_style = dict(color='black', linewidth=1.2)

    ax.add_patch(patches.FancyBboxPatch((40, 85), 30, 8, boxstyle="round,pad=0.2", **root_style))
    ax.text(55, 89, "Sistemas de\n Aprendizaje Automático", color='white', ha='center', va='center', fontsize=14, weight='bold')

    data = [
        (5, "Según la\nsupervisión", ["Supervisado", "No supervisado", "Por refuerzo"]),
        (42, "Según flujo\nde datos", ["Aprendizaje Batch", "Aprendizaje Online"]),
        (78, "Según cómo\ngeneralizan", ["Por instancias", "Basado en modelo"])
    ]

    for x_cat, name, children in data:
        box_width = 25
        ax.add_patch(patches.FancyBboxPatch((x_cat, 65), box_width, 8, boxstyle="round,pad=0.2", **cat_style))
        ax.text(x_cat + (box_width/2), 69, name, ha='center', va='center', fontsize=13, weight='bold')

        ax.annotate('', xy=(x_cat + (box_width/2), 73.5), xytext=(55, 85), 
                    arrowprops=dict(arrowstyle='->', connectionstyle="angle,angleA=0,angleB=90,rad=5"))

        v_line_x = x_cat + 3
        y_top = 65
        y_bottom = 50 - (len(children)-1) * 10
        ax.plot([v_line_x, v_line_x], [y_top, y_bottom + 3], **line_style)

        for i, child_name in enumerate(children):
            y_pos = 50 - (i * 10)
            child_box_width = 19
            
            ax.annotate('', xy=(v_line_x + 2, y_pos + 3), xytext=(v_line_x, y_pos + 3),
                        arrowprops=dict(arrowstyle='->', color='black'))
            
            # Rectángulo del hijo
            ax.add_patch(patches.Rectangle((v_line_x + 2, y_pos), child_box_width, 6, **sub_style))
            ax.text(v_line_x + 2 + (child_box_width/2), y_pos + 3, child_name, ha='center', va='center', fontsize=12)

    plt.tight_layout()
    plt.show()

draw_ml_tree_final()
```
