```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

def draw_rnn_final_version():
    fig, ax = plt.subplots(figsize=(14, 6))
    # Ajustamos límites para que el bucle más ancho no se corte
    ax.set_xlim(-3, 17) 
    ax.set_ylim(-1, 6)
    ax.set_aspect('equal')
    ax.axis('off')

    # Configuración de estilo
    box_w, box_h = 1.5, 0.8
    purple_box = {'facecolor': "#dabbdf", 'edgecolor': 'black', 'linewidth': 1.2}
    blue_box = {'facecolor': '#98c1d9', 'edgecolor': 'black', 'linewidth': 1.2}

    # --- PARTE (A): ---
    ax.add_patch(patches.Rectangle((0, 0), box_w, box_h, **blue_box))
    ax.add_patch(patches.Rectangle((0, 2), box_w, box_h, **purple_box))
    ax.add_patch(patches.Rectangle((0, 4), box_w, box_h, **purple_box))

    ax.text(0.75, 0.4, r'$\mathbf{x}$', ha='center', va='center', fontsize=12)
    ax.text(0.75, 2.4, r'$\mathbf{h}$', ha='center', va='center', fontsize=12)
    ax.text(0.75, 4.4, r'$\mathbf{y}$', ha='center', va='center', fontsize=12)

    ax.arrow(0.75, 0.8, 0, 1.1, head_width=0.15, fc='black', length_includes_head=True)
    ax.text(1.2, 1.4, '$W_{x,h}$', fontsize=11)
    ax.arrow(0.75, 2.8, 0, 1.1, head_width=0.15, fc='black', length_includes_head=True)
    ax.text(1.2, 3.4, '$W_{h,y}$', fontsize=11)

    loop = patches.Arc((0, 2.4), 1.8, 1.2, theta1=90, theta2=270, color='black', linewidth=1.2)
    ax.add_patch(loop)
    ax.arrow(0, 1.8, 0.1, 0, head_width=0.15, fc='black', length_includes_head=True) 
    
    ax.text(-1.1, 2.4, 'Delay', ha='right', va='center', style='italic', fontsize=10)
    ax.text(-0.7, 1.6, '$W_{h,h}$', ha='center', fontsize=11)
    
    ax.text(0.75, -0.8, '(a)', fontsize=14, ha='center', weight='bold')

    # --- PARTE (B): UNROLLED ---
    h0_x = 4
    ax.add_patch(patches.Rectangle((h0_x, 2), box_w, box_h, **purple_box))
    ax.text(h0_x + 0.75, 2.4, r'$\mathbf{h_0}$', ha='center', va='center', fontsize=11)

    first_step_x = h0_x + 2.8  
    
    ax.arrow(h0_x + 1.5, 2.4, 1.2, 0, head_width=0.15, fc='black', length_includes_head=True)
    ax.text(h0_x + 2.1, 2.8, '$W_{h,h}$', ha='center', fontsize=10)

    for t in range(1, 4):
        curr_x = first_step_x + (t-1) * 3
        
        ax.add_patch(patches.Rectangle((curr_x, 0), box_w, box_h, **blue_box))
        ax.add_patch(patches.Rectangle((curr_x, 2), box_w, box_h, **purple_box))
        ax.add_patch(patches.Rectangle((curr_x, 4), box_w, box_h, **purple_box))
        
        ax.text(curr_x + 0.75, 0.4, f'$\\mathbf{{x_{t}}}$', ha='center', va='center', fontsize=11)
        ax.text(curr_x + 0.75, 2.4, f'$\\mathbf{{h_{t}}}$', ha='center', va='center', fontsize=11)
        ax.text(curr_x + 0.75, 4.4, f'$\\mathbf{{y_{t}}}$', ha='center', va='center', fontsize=11)

        ax.arrow(curr_x + 0.75, 0.8, 0, 1.1, head_width=0.15, fc='black', length_includes_head=True)
        ax.text(curr_x + 1.2, 1.4, '$W_{x,h}$', fontsize=9)
        ax.arrow(curr_x + 0.75, 2.8, 0, 1.1, head_width=0.15, fc='black', length_includes_head=True)
        ax.text(curr_x + 1.2, 3.4, '$W_{h,y}$', fontsize=9)

        if t < 3:
            ax.arrow(curr_x + 1.5, 2.4, 1.4, 0, head_width=0.15, fc='black', length_includes_head=True)
            ax.text(curr_x + 2.2, 2.8, '$W_{h,h}$', ha='center', fontsize=9)

    ax.text(first_step_x + 3.75, -0.8, '(b)', fontsize=14, ha='center', weight='bold')

    plt.tight_layout()
    plt.show()

draw_rnn_final_version()
```
